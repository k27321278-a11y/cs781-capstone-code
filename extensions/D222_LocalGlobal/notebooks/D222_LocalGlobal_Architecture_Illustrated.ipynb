## Architecture

![Path](../figures/architecture.svg)

A 24-block, width-896 causal decoder uses 14 query heads over two shared key/value heads. Every fourth block attends globally and intervening blocks use 512-token local attention. The 222M-scale run was built after the original Phase 1 deadline.

## Tensor mechanics

A token ID batch `[B,T]` is embedded into width `d`; attention projects queries, keys and values into heads, applies causal or encoder masks, and adds a residual update. RMSNorm scales activations without subtracting their mean. SwiGLU gates a feed-forward branch. Rotary phases encode relative position within query/key vectors. The final tied embedding matrix maps hidden states back to vocabulary logits. The B family adds a bidirectional prompt encoder and cross-attention; the C family replaces the token decoder stack with a GRU; LegalSpan adds prompt routing and a copy gate.

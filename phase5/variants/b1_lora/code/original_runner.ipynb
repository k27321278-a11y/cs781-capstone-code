```python
import os, subprocess
from pathlib import Path
roots = list(Path("/kaggle/input").rglob("phase5_b1_train.py"))
print("Mounted runners:", [str(p) for p in roots], flush=True)
input_dir = str(roots[0].parent) if len(roots) == 1 else None
if input_dir is None:
    raise RuntimeError("Expected exactly one mounted official-data runner")
output_dir = "/kaggle/working"
command = ["python", f"{input_dir}/phase5_b1_train.py", "--input", input_dir, "--output", output_dir, "--epochs", "2", "--batch", "8", "--eval-batch", "16", "--accum", "2", "--lr", "0.0002"]
print("Running own Phase 1 B1 LoRA classifier on official Phase 5 LSI data", flush=True)
subprocess.run(command, check=True, env={**os.environ, "PYTHONPATH": input_dir + "/phase1_source:" + input_dir})
```

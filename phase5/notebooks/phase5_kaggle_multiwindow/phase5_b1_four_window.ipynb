```python
import os, sys, json
from pathlib import Path
from types import SimpleNamespace
sources = list(Path("/kaggle/input").rglob("phase5_multiwindow.py"))
print("Mounted four-window sources:", [str(p) for p in sources], flush=True)
if len(sources) != 1: raise RuntimeError("Expected one validated four-window source")
input_dir = str(sources[0].parent)
sys.path[:0] = [input_dir + "/phase1_source", input_dir]
import phase5_b1_train as runner
import phase5_multiwindow as multi
multi.install()
output_dir = Path("/kaggle/working/four_window_adapter")
args = SimpleNamespace(input=input_dir, output=str(output_dir), epochs=4, batch=8, eval_batch=16, accum=2, lr=0.0002)
runner.train(args)
report_file = output_dir / "phase5_b1_lora_report.json"
report = json.loads(report_file.read_text())
report["method"] = "B1 four-window bottleneck adapters with label-wise smooth-max evidence pooling"
report["window_count"] = 4
report["tokens_per_window"] = 192
report["source_reference"] = "https://aclanthology.org/D19-1044/"
(output_dir / "phase5_b1_four_window_report.json").write_text(json.dumps(report, indent=2))
(output_dir / "phase5_b1_lora_prediction.zip").rename(output_dir / "phase5_b1_four_window_prediction.zip")
```
